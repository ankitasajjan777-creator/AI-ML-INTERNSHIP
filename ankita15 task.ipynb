```python
# Sample Space for two consecutive customer actions

actions = ["Click", "Scroll", "Exit"]

sample_space = [(a1, a2) for a1 in actions for a2 in actions]

sample_space
```

```python
# Filter outcomes where "Click" appears at least once

event_click_once = [pair for pair in sample_space if "Click" in pair]

# Calculate probability
prob_click_once = len(event_click_once) / len(sample_space)

print("Sample Space:", sample_space)
print("\nEvent (Clicks at least once):", event_click_once)
print("\nProbability of clicking at least once:", prob_click_once)
```

```python
import random

count_7 = 0
trials = 1000

for _ in range(trials):
    d1 = random.randint(1, 6)
    d2 = random.randint(1, 6)
    if d1 + d2 == 7:
        count_7 += 1

probability = count_7 / trials
print("Experimental Probability of sum = 7:", probability)
```

```python
# Independent events: Coin flip AND die roll
P_heads = 1 / 2
P_six = 1 / 6
P_independent = P_heads * P_six

# Dependent events: Two red marbles without replacement
# Bag: 5 Red, 5 Blue
P_first_red = 5 / 10
P_second_red_given_first = 4 / 9
P_dependent = P_first_red * P_second_red_given_first

print("P(Heads AND 6) =", P_independent)
print("P(Both Red marbles) =", P_dependent)
```

```python
# Given probabilities
P_spam = 0.10
P_ham = 0.90
P_free_given_spam = 0.90
P_free_given_ham = 0.05

# Total probability of seeing "Free"
P_free = (P_free_given_spam * P_spam) + (P_free_given_ham * P_ham)

# Bayes' Theorem
P_spam_given_free = (P_free_given_spam * P_spam) / P_free

print("P(Free) =", P_free)
print("P(Spam | Free) =", P_spam_given_free)
```
