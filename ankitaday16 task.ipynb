```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
heights = pd.Series(np.random.normal(loc=170, scale=10, size=1000))
```

```python
income = pd.Series(np.random.exponential(scale=50000, size=1000))
```

```python
scores = pd.Series(100 - np.random.exponential(scale=10, size=1000))
```

```python
plt.figure(figsize=(18,5))
```

```python
plt.subplot(1,3,1)
sns.histplot(heights, kde=True)
plt.title("Human Heights (Normal Distribution)")
```

```python
plt.subplot(1,3,2)
sns.histplot(income, kde=True)
plt.title("Household Income (Right-Skewed)")
```

```python
plt.subplot(1,3,3)
sns.histplot(scores, kde=True)
plt.title("Test Scores (Left-Skewed)")

plt.tight_layout()
plt.show()
```

```python
datasets = {
    "Human Heights": heights,
    "Household Income": income,
    "Test Scores": scores
}

for name, data in datasets.items():
    mean_value = data.mean()
    median_value = data.median()

    print(f"\nDataset: {name}")
    print(f"Mean   : {mean_value:.2f}")
    print(f"Median : {median_value:.2f}")

     # Observation
    if mean_value > median_value:
        print("Observation: Right-Skewed Distribution")
    elif mean_value < median_value:
        print("Observation: Left-Skewed Distribution")
    else:
        print("Observation: Approximately Normal Distribution")

   
```

```python
## Explanation of Task 1 – The Shape Shifter #

In this task, three different types of data distributions were analyzed: Normal, Right-Skewed, and Left-Skewed.

First, numerical datasets were generated using Pandas:

Human Heights were generated using a Normal distribution.

Household Incomes were generated using an Exponential distribution to create a Right-Skewed pattern.

Test Scores for an easy exam were generated to create a Left-Skewed pattern.

Next, Histograms with KDE (Kernel Density Estimate) overlays were plotted to visually understand the shape of each distribution.

Then, the Mean and Median were calculated for each dataset to identify skewness:

For Human Heights, the Mean and Median are approximately equal, indicating a Normal distribution.

For Household Income, the Mean is greater than the Median, indicating a Right-Skewed distribution. This happens because a few high-income values pull the mean toward the right.

For Test Scores, the Mean is less than the Median, indicating a Left-Skewed distribution. This occurs because most students scored high, with a few low scores pulling the mean downward.

According to the observation rule:

If Mean > Median → Right-Skewed

If Mean < Median → Left-Skewed

Understanding skewness is important because many Machine Learning algorithms assume normally distributed data. If the data is skewed, it may affect model performance and require transformation.

If you want, I can also give you a slightly more professional version to impress your mento
```

```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)
```

```python
data = pd.DataFrame({
    "SAT_Score": np.random.normal(loc=1000, scale=100, size=100)
})
```

```python
data.loc[100] = 1500
data.loc[101] = 400
```

```python
mu = data["SAT_Score"].mean()
sigma = data["SAT_Score"].std()

print("Mean (μ):", round(mu, 2))
print("Standard Deviation (σ):", round(sigma, 2))
```

```python
data["z_score"] = (data["SAT_Score"] - mu) / sigma
```

```python
outliers = data[np.abs(data["z_score"]) > 3]

print("\nOutliers (|Z| > 3):")
print(outliers)
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as pl
```

```python
np.random.seed(42)
```

```python
population = np.random.exponential(scale=50, size=10000)
```

```python
plt.figure(figsize=(12,5))
```

```python
plt.subplot(1,2,1)
sns.histplot(population, kde=True)
plt.title("Original Skewed Population")
```

```python

sample_means = []
for i in range(1000):
    sample = np.random.choice(population, size=30)
    sample_means.append(np.mean(sample))

sample_means = pd.Series(sample_means)
```

```python

plt.subplot(1,2,2)
sns.histplot(sample_means, kde=True)
plt.title("Distribution of 1000 Sample Means")

plt.tight_layout()
plt.show()
```

```python
print("Mean of Population:", round(np.mean(population), 2))
print("Mean of Sample Means:", round(sample_means.mean(), 2))
```
