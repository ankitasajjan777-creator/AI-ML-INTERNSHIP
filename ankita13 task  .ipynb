```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample housing data
data = {
    'SquareFootage': [800, 1000, 1200, 1500, 1800, 2000, 2500],
    'Price': [200000, 250000, 300000, 360000, 420000, 450000, 520000],
    'Location': ['Urban', 'Urban', 'Suburban', 'Suburban', 'Suburban', 'Urban', 'Urban']
}

df = pd.DataFrame(data)
```

```python
plt.figure()
sns.scatterplot(x='SquareFootage', y='Price', data=df)
plt.title('Square Footage vs Price')
plt.xlabel('Square Footage')
plt.ylabel('Price')
plt.show()
```

```python
plt.figure()
sns.boxplot(x='Location', y='Price', data=df)
plt.title('Price Distribution by Location')
plt.xlabel('Location')
plt.ylabel('Price')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample dataset
data = {
    'SquareFootage': [800, 1000, 1200, 1500, 1800, 2000, 2500],
    'Price': [200000, 250000, 300000, 360000, 420000, 450000, 520000],
    'Bedrooms': [2, 2, 3, 3, 4, 4, 5]
}

df = pd.DataFrame(data)

corr_matrix = df.corr()
print(corr_matrix)
```

```python
plt.figure()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix Heatmap')
plt.show()
```

```python
plt.figure()
sns.boxplot(y='Price', data=df)
plt.title('Boxplot of Price')
plt.show()
```
